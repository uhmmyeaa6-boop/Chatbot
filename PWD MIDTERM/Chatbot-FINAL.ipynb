### 2.1 Mapping inputs to responses using dictionaries.

We now use dictionaries to map user intents. Instead of writing a new *dictionary*, use the **ZIP** function on a list. 

```python
# Write a function that takes in precept categories and responses lists and returns a mapping between them.

# - Use ZIP function to combine two independent lists (precepts and responses) and create a new map between them.
# - Search on the ZIP function and how to use it in a for loop
# - Run the main loop and check that this function works

# ___ Write your code here _____
# Function to create a dictionary mapping using ZIP
def zip_responses(precepts, responses):
    mapping = {}
    for precept, response in zip(precepts, responses):
        mapping[precept] = response
    return mapping
```

### 2.2 Using regular expressions to match patterns.

- Match complete words using regular expressions. 
- Example, match all occurrences of hello, hi, hey etc under a pattern called greetings.
- Use "|" symbol to string together multiple words.
- Each pattern should have atleast or 5 different occurrences of the word you want to match.
- Create atleast 5 different patterns.

---

```python
# Write a function to create 5 RE patterns and return them as separate variables.
# Give the function a characteristic name (e.g., re_patterns) 
# ___ Write your code here _____
import re
import random

# Function 
def re_patterns():
    greetings = r"\b(hello|hi|hey|greetings)\b"
    bye = r"\b(bye|goodbye|see you|byebye|cyaa)\b"
    thanks = r"\b(thank|thanks|thank you|thankyou|appreciate|grateful|cheers)\b"
    bookish = r"\b(book|books|novel|novels|find|search|look|looking|recommend|suggest)\b"
    author = r"\b(author|authors|writer|written|who\s+wrote|by|authored)\b"
    genre = r"\b(genre|genres|type|types|kind|category|categories)\b"
    title = r"\b(title|name|named|called|book\s+called|book\s+named)\b"
    dostoevsky = r"\b(dostoevsky|fyodor dostoyevsky|russian literature)\b"
    crime_and_punishment = r"\b(crime and punishment|crime & punishment)\b"
    horror = r"\b(horror|scary|spooky|creepy|thriller)\b"
    
    return greetings, bye, thanks, bookish, author, genre, title, dostoevsky, crime_and_punishment, horror
```

Follow the steps to write function a function to detect patterns in the user's input. The function returns a response type (e.g., greetings, bye).

To detect a pattern from user input, you will use regular expressions created before.



```python
## Detect pattern - Write a function to detect RE patterns in user input and return response category.

## Create the detect_pattern function with one argument : user_input
## 1. Call the re_patterns function and store patterns in seperate variables
## 2. Write if-else conditions to check for pattern match
##    2.1 For the if condition, use the re.search to check if the pattern exists in user input 
##    2.2 If there is a match, return the response type (e.g., "greetings" or "bye" or "happy")
##    2.3 Note: Use <span style="color:red; font-size: 14px"> elif </span> to match multiple conditions
##    2.4 If none of the conditions, return "default" which will be the default response of the chatbot
##    2.5 <span style="color:red; font-size: 14px"> Note: use else for last conditon </span>


# Give the function a characteristic name (e.g., detect_pattern) 
# ___ Write your code here _____
def detect_pattern(user_input):
    greetings, bye, thanks, bookish, author, genre, title, dostoevsky, crime_and_punishment, horror = re_patterns()
    
    # for greetings
    if re.search(greetings, user_input, re.IGNORECASE):
        return "greetings"
    # for bye
    elif re.search(bye, user_input, re.IGNORECASE):
        return "bye"
    # thanks
    elif re.search(thanks, user_input, re.IGNORECASE):
        return "thanks"
    # bookish terms
    elif re.search(bookish, user_input, re.IGNORECASE):
        return "bookish"
    # author
    elif re.search(author, user_input, re.IGNORECASE):
        return "author"
    #  genre
    elif re.search(genre, user_input, re.IGNORECASE):
        return "genre"
    # title
    elif re.search(title, user_input, re.IGNORECASE):
        return "title"
    # specific authors/books
    elif re.search(dostoevsky, user_input, re.IGNORECASE):
        return "dostoevsky"
    elif re.search(crime_and_punishment, user_input, re.IGNORECASE):
        return "crime_and_punishment"
    elif re.search(horror, user_input, re.IGNORECASE):
        return "horror"
    else:
        return "default"
```

```python
## Chatbot response - Write a function to return a responses by indexing dictionary with response categories.

## Create the chatbot_response function with one argument : user_input
## 1. Call the detect_pattern function and store the response in variable (e.g., detected_type)
## 2. Call the generate_responses function and store responses in a variable (e.g., re_response)
## 2. Write if-else conditions to check for pattern match
##    2.1 For the if condition, check if the detected_type is one of the keys of the dictionary
##    2.2 If there is a match, return the response from dictonary
##    2.3 Note: Use elif to match multiple conditions
##    2.4 If none of the conditions, use "default" response

# Give the function a characteristic name (e.g., detect_pattern) 
# ___ Write your code here _____
def generate_responses():
    responses = {
        "greetings": ["Hi there! How are you doing today?", "Hello! How can I help you?", "Hey! What can I do for you today?"],
        "bye": ["Goodbye! Have a great day.", "See you later!", "Bye! Come back soon!"],
        "thanks": ["You are welcome!", "Happy to help!", "Anytime!"],
        "bookish": ["Sure, what book are you looking for? You can search by author, title or genre.", "I'd be happy to help you find a book! What are you interested in?"],
        "author": ["Please enter the author's name.", "Which author are you looking for?"],
        "genre": ["Please enter the genre.", "What genre interests you?"],
        "title": ["Please enter the title.", "What's the book title you're looking for?"],
        "dostoevsky": ["Dostoevsky: Crime and Punishment, The Brothers Karamazov is available to borrow."],
        "crime_and_punishment": ["Crime and Punishment is available to borrow."],
        "horror": ["Horror: The Little Cat is available to borrow."],
        "default": ["I'm sorry, I didn't understand that.", "Could you please rephrase that?", "I'm not sure what you mean. Can you try again?"]
    }
    
    return responses
def chatbot_response(user_input):
    # call function and store response type
    detected_type = detect_pattern(user_input)
    
    # Call function and store responses dictionary
    responses = generate_responses()
    
    # Check if detected_type is in responses dictionary
    if detected_type in responses:
        # Return any response from the list
        return random.choice(responses[detected_type])
    else:
        # Return default response
        return random.choice(responses["default"])
```

#### Writing advanced regular expressions

Let's exploring advanced REs by creating two more patterns to match the name and favourite color of the user. 

As name, and color can be any value, pattern matching needs to be generic and requires using alphabets [a-zA-Z], words [\w] and special symbols [*,+,?].

```python
# Write a regular expression pattern for a name and favourite color.

# 1. Names typically start with a letter in upper case followed by a sequence of letters in lower case.
# Try writing a few use cases to ensure that the pattern works.

# 2. Colors typically start with a pattern - color is <color_name> or I like <color_name>.
# Try writing a few use cases to ensure that the pattern works.
import re

def get_response(user_input, pattern_dict):
    for pattern, response in pattern_dict.items():
        if re.search(pattern, user_input, re.IGNORECASE):
            return response
    return "I'm sorry, I didn't understand that."

#  pattern dictionary for responses
pattern_dict = {
    r"\b(hello|hi|hey|hiya|greetings)\b": "Hi there! How are you doing today?",
    r"\b(how are you|how's it going)\b": "I'm doing great, thanks for asking!",
    r"\b(thank|thanks|thank you)\b": "You're welcome!",
    r"\b(bye|goodbye|see you)\b": "Goodbye! Have a great day."
}

state = {}
dynamic_patterns = {
    r"(?:my name is|i'm|i am|call me)\s+([A-Z][a-z]+(?:\s+[A-Z][a-z]+)*)": "Nice to meet you, {name}!",
    r"(?:my favorite color is|my favourite color is|i like|color is|colour is)\s+([a-z]+)": "{color} is a beautiful color!"
}

print("Bot: Hello! I'm here to chat with you. Tell me your name or favorite color!")

while True:
    user_input = input("You: ")
    
    # Check patterns first (name and color)
    for pattern, template in dynamic_patterns.items():
        match = re.search(pattern, user_input, re.IGNORECASE)
        if match:
            if "name" in template:
                state['name'] = match.group(1)
                print("Bot:", template.format(name=state['name']))
            elif "color" in template:
                state['color'] = match.group(1)
                print("Bot:", template.format(color=state['color']))
            break
    else:
        # Fallback to predefined pattern_dict
        response = get_response(user_input, pattern_dict)
        print("Bot:", response)
        if re.search(r"\b(bye|goodbye)\b", user_input, re.IGNORECASE):
            break
```

```python
import json
import random
import nltk
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.tokenize import word_tokenize

# Download  packages
nltk.download('punkt_tab', quiet=True)
nltk.download('wordnet', quiet=True)
nltk.download('omw-1.4', quiet=True)

# Initialize stemmer and lemmatizer
ps = PorterStemmer()
wnl = WordNetLemmatizer()


def load_intents():
    
    with open('intents.json', 'r') as file:
        data = json.load(file)
    return data['intents']


def find_match(user_input, intents):
   
    
    # Tokenize user input
    user_tokens = word_tokenize(user_input.lower())
    
    # Check each intent
    for intent in intents:
        for pattern in intent['patterns']:
            # Tokenize pattern
            pattern_tokens = word_tokenize(pattern.lower())
            
            # Try stemming match
            for user_word in user_tokens:
                user_stem = ps.stem(user_word)
                for pattern_word in pattern_tokens:
                    pattern_stem = ps.stem(pattern_word)
                    if user_stem == pattern_stem:
                        return intent['tag'], intent['responses']
            
            # Try lemmatization match
            for user_word in user_tokens:
                user_lemma = wnl.lemmatize(user_word, pos='v')
                for pattern_word in pattern_tokens:
                    pattern_lemma = wnl.lemmatize(pattern_word, pos='v')
                    if user_lemma == pattern_lemma:
                        return intent['tag'], intent['responses']
    
    return None, None


def chatbot():
    
    intents = load_intents()
    
    print("\n" + "="*50)
    print("Library Chatbot")
    print("Type 'bye' to exit")
    print("="*50 + "\n")
    
    while True:
        user_input = input("You: ")
        
        # Check for exit
        if user_input.lower() in ['bye', 'exit', 'quit']:
            print("Chatbot: Goodbye! Have a great day!")
            break
        
        # Find match
        tag, responses = find_match(user_input, intents)
        
        if responses:
            response = random.choice(responses)
            print(f"Chatbot: {response}\n")
        else:
            print("Chatbot: I'm sorry, could you rephrase that?\n")


# Run the chatbot
chatbot()
```

```python
##adv feature 
import json
import random
import re

# Max books a user can borrow
MAX_BOOKS = 3

def load_intents():
    with open('intents.json', 'r') as file:
        data = json.load(file)
    return data['intents']

def find_match(user_input, intents):
    user_input = user_input.lower()
    
    for intent in intents:
        for pattern in intent['patterns']:
            if pattern.lower() in user_input:
                return intent['responses']
    return None

def borrow_book(book_name, borrowed_books):
    # Check if limit reached
    if len(borrowed_books) >= MAX_BOOKS:
        return f"Sorry! You already borrowed {MAX_BOOKS} books. Return one first!"
    
    # Check if already borrowed
    if book_name in borrowed_books:
        return f"You already borrowed '{book_name}'!"
    
    # Borrow the book
    borrowed_books.append(book_name)
    left = MAX_BOOKS - len(borrowed_books)
    return f"Done! You borrowed '{book_name}'. You can borrow {left} more."

def return_book(book_name, borrowed_books):
    if book_name in borrowed_books:
        borrowed_books.remove(book_name)
        left = MAX_BOOKS - len(borrowed_books)
        return f"Thanks for returning '{book_name}'! You can borrow {left} more."
    else:
        return f"You didn't borrow '{book_name}'."

def show_books(borrowed_books):
    if not borrowed_books:
        return "You haven't borrowed any books."
    
    books = ", ".join(borrowed_books)
    left = MAX_BOOKS - len(borrowed_books)
    return f"Your books: {books}. You can borrow {left} more."

def chatbot():
    intents = load_intents()
    borrowed_books = []
    
    print("\n" + "="*50)
    print("Library Chatbot - Borrow Limit System")
    print(f"Max books: {MAX_BOOKS}")
    print("="*50)
   
    while True:
        user_input = input("You: ")
        
        # Check for borrow
        if user_input.lower().startswith('borrow '):
            book = user_input[7:].strip()
            response = borrow_book(book, borrowed_books)
            print(f"Bot: {response}\n")
            continue
        
        # Check for return
        if user_input.lower().startswith('return '):
            book = user_input[7:].strip()
            response = return_book(book, borrowed_books)
            print(f"Bot: {response}\n")
            continue
        
        # Check for my books
        if 'my books' in user_input.lower():
            response = show_books(borrowed_books)
            print(f"Bot: {response}\n")
            continue
        
        # Regular chat
        responses = find_match(user_input, intents)
        if responses:
            print(f"Bot: {random.choice(responses)}\n")
        else:
            print("Bot: I'm sorry, I didn't understand that.\n")
        
        # Exit
        if 'bye' in user_input.lower():
            break

# Run
chatbot()
```
